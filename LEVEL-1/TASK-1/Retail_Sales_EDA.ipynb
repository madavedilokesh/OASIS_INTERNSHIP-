Creating Month Column

```python
df['Month'] = df['Date'].dt.to_period('M')
```

```python
df.columns
```

```python
monthly_sales = df.groupby('Month')['Total Amount'].sum()
```

```python
monthly_sales
```

**Time Series Analysis**

**1.Monthly Sales**

```python
monthly_sales.plot(kind='line', marker='o', figsize=(12,5))

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Amount')
plt.grid(True)
plt.show()
```

Description:
The monthly sales trend shows how the total sales amount changes from month to month. It helps identify periods of increasing or decreasing sales, monthly fluctuations, and possible seasonal patterns in customer purchasing behavior.The sales are high in the month of may .

2.Quarterly Sales

```python
df['Quarter'] = df['Date'].dt.to_period('Q')
```

```python
quarterly_sales = df.groupby('Quarter')['Total Amount'].sum()
```

```python
quarterly_sales
```

```python
quarterly_sales.plot(kind='line', marker='o', figsize=(10,5))

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Amount')
plt.grid(True)
plt.show()
```

Description:
The quarterly sales trend summarizes total sales for each quarter of the year. It provides a broader view of sales performance and helps identify which quarters perform better or worse, making it easier to understand overall business growth and seasonal trends.The sales are high in the 4th Quarter(Q4) in 2023.

**Customer Demographics Analysis**

**1.Age Groups**

```python
df['Age Group'] = pd.cut(
    df['Age'],
    bins=[0, 18, 25, 35, 45, 55, 100],
    labels=['<18', '18-25', '26-35', '36-45', '46-55', '56+']
)
```

```python
df[['Age', 'Age Group']].head()
```

```python
age_counts = df.groupby('Age Group', observed=True)['Customer ID'].nunique()
```

```python
age_counts
```

```python
age_counts.plot(kind='bar', figsize=(8,5))

plt.title('Customer Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)

plt.show()
```

Description : This plot shows the distribution of customers across different age groups. It helps identify which age group has the highest number of customers and provides an overview of the age profile of the customer base. In the 46-55 age group the customers are more.and below 18 the customers are very low.

**2.Gender**

```python
gender_counts = df.groupby('Gender')['Customer ID'].nunique()
```

```python
gender_counts
```

```python
gender_counts.plot(kind='bar', figsize=(7,5))

plt.title('Customer Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)

plt.show()
```

Description : This plot shows the distribution of unique customers by gender. It helps compare the number of customers across different genders and understand the overall gender composition of the customer base.The most customers are Female only and the number is almost 500+ customers.

**Product Analysis**

**1. Top 10 best selling products**

```python
top_10_categories = (
    df.groupby('Product Category')['Quantity']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
```

```python
top_10_categories
```

```python
top_10_categories.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Top 10 Best-Selling Product Categories')
plt.xlabel('Product Category')
plt.ylabel('Total Quantity Sold')
plt.xticks(rotation=45)
plt.show()
```

Description : This chart shows the top 10 product categories based on the total quantity sold. It helps identify the categories with the highest customer demand and highlights the best-performing categories in terms of units sold.The top category that sold is clothing most customers will choose the clothing.

**2. Revenue by Product Category**

```python
revenue_category = (
    df.groupby('Product Category')['Total Amount']
      .sum()
      .sort_values(ascending=False)
)
```

```python
revenue_category
```

```python
revenue_category.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

Description : This chart shows the total revenue generated by each product category. It helps identify which categories contribute the most to overall revenue and provides insights into the financial performance of different product categories.Even though the most saled  category is clothing but the highest revenue earned category is electronics. Thats why we prefer electronics stock more.

**Heat Map**

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    df[num_cols].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
```

Description :The correlation heatmap shows the strength and direction of relationships between the numerical variables in the dataset. It helps identify strongly related variables and provides insights into how changes in one numerical variable are associated with changes in another.

Additional Visualization

Product Category vs Gender

```python
category_gender_sales = df.groupby(
    ['Product Category', 'Gender']
)['Total Amount'].sum().unstack()
```

```python
category_gender_sales
```

```python
category_gender_sales.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Revenue by Product Category and Gender')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.legend(title='Gender')
plt.show()
```

Description:This visualization compares revenue across product categories by gender. It can reveal differences in purchasing patterns between genders and identify product categories where one gender contributes significantly more revenue than the other.
